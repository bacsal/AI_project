```python
def get_challenger_name():
    list = []
    api_url = ("https://" +
               "kr.api.riotgames.com/lol/league-exp/v4/entries/RANKED_SOLO_5x5/CHALLENGER/"+
               "I"+
               "?page=1"+
               "&api_key="+
               api_key
               )
    resp = requests.get(api_url)
    challenger_data = resp.json()
    for challenger_info in challenger_data:
        challenger_name = challenger_info['summonerName']
        list.append(challenger_name) 
    return list
```

```python
def get_match_data(region, match_id, api_key):
    api_url =( 
        "https://" + 
        region + 
        ".api.riotgames.com/lol/match/v5/matches/" +
        match_id +
        "?api_key=" +
        api_key 
    )
    resp = requests.get(api_url)
    data = resp.json()
    return data
```

```python
def did_win(puuid, match_data):
    part_index = match_data['metadata']['participants'].index(puuid)
    return match_data['info']['participants'][part_index]['win']
```

```python
def get_puuid(SummonerNameList):
    puuid_list = []
    for name in SummonerNameList:
        api_url = (
            "https://kr.api.riotgames.com/lol/summoner/v4/summoners/by-name/" +
            name +
            "?api_key="+
            api_key
        )
        resp = requests.get(api_url)
        Summoner_data = resp.json()
        print(Summoner_data['puuid'])
    return puuid_list
```
